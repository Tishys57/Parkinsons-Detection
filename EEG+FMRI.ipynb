```python
import numpy as np
import pandas as pd
import nibabel as nib
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils import class_weight
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.ndimage import zoom
import warnings
import optuna
import os
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

class EEGNet:
    """
    EEGNet with added layers and regularization
    """
    def __init__(self, nb_classes=2, chans=64, samples=891, dropoutRate=0.5, 
                 kernLength=64, F1=16, D=2, F2=32, dropoutType='Dropout'):
        self.nb_classes = nb_classes
        self.chans = chans
        self.samples = samples
        self.dropoutRate = dropoutRate
        self.kernLength = kernLength
        self.F1 = F1
        self.D = D
        self.F2 = F2
        self.dropoutType = dropoutType
        
    def build_model(self, input_shape):
        input_layer = layers.Input(shape=input_shape)
        reshaped = layers.Reshape((self.chans, self.samples, 1))(input_layer)
        
        # Block 1
        block1 = layers.Conv2D(self.F1, (1, self.kernLength), padding='same', use_bias=False,
                              kernel_regularizer=regularizers.l2(0.001))(reshaped)
        block1 = layers.BatchNormalization()(block1)
        block1 = layers.DepthwiseConv2D((self.chans, 1), use_bias=False, 
                                       depth_multiplier=self.D,
                                       depthwise_constraint=tf.keras.constraints.max_norm(1.))(block1)
        block1 = layers.BatchNormalization()(block1)
        block1 = layers.Activation('elu')(block1)
        block1 = layers.AveragePooling2D((1, 4))(block1)
        block1 = layers.Dropout(self.dropoutRate)(block1)
        
        # Block 2
        block2 = layers.SeparableConv2D(self.F2, (1, 16), use_bias=False, padding='same')(block1)
        block2 = layers.BatchNormalization()(block2)
        block2 = layers.Activation('elu')(block2)
        block2 = layers.AveragePooling2D((1, 8))(block2)
        block2 = layers.Dropout(self.dropoutRate)(block2)
        
        # Additional Block
        block3 = layers.SeparableConv2D(self.F2*2, (1, 8), use_bias=False, padding='same')(block2)
        block3 = layers.BatchNormalization()(block3)
        block3 = layers.Activation('elu')(block3)
        block3 = layers.AveragePooling2D((1, 4))(block3)
        block3 = layers.Dropout(self.dropoutRate)(block3)
        
        flatten = layers.Flatten(name='features')(block3)
        attention = layers.Dense(128, activation='relu')(flatten)
        attention = layers.Dropout(self.dropoutRate)(attention)
        
        if self.nb_classes == 2:
            dense = layers.Dense(1, name='predictions', kernel_constraint=tf.keras.constraints.max_norm(0.25))(attention)
            output = layers.Activation('sigmoid', name='output')(dense)
            loss = 'binary_crossentropy'
        else:
            dense = layers.Dense(self.nb_classes, name='predictions', kernel_constraint=tf.keras.constraints.max_norm(0.25))(attention)
            output = layers.Activation('softmax', name='output')(dense)
            loss = 'sparse_categorical_crossentropy'
        
        model = models.Model(inputs=input_layer, outputs=output)
        model.compile(loss=loss, optimizer=tf.keras.optimizers.AdamW(learning_rate=0.001), metrics=['accuracy'])
        return model

class ShallowCNN:
    """
    Enhanced ShallowCNN with deeper architecture
    """
    def __init__(self, input_shape, num_filters=64, kernel_size=(3, 3), 
                 pool_size=(2, 2), dropout_rate=0.4):
        self.input_shape = input_shape
        self.num_filters = num_filters
        self.kernel_size = kernel_size
        self.pool_size = pool_size
        self.dropout_rate = dropout_rate
        
    def build_model(self):
        input_layer = layers.Input(shape=self.input_shape)
        
        x = layers.Conv2D(self.num_filters, self.kernel_size, activation='relu', padding='same',
                         kernel_regularizer=regularizers.l2(0.001))(input_layer)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling2D(self.pool_size)(x)
        x = layers.Dropout(self.dropout_rate)(x)
        
        x = layers.Conv2D(self.num_filters * 2, self.kernel_size, activation='relu', padding='same',
                         kernel_regularizer=regularizers.l2(0.001))(x)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling2D(self.pool_size)(x)
        x = layers.Dropout(self.dropout_rate)(x)
        
        x = layers.Conv2D(self.num_filters * 4, self.kernel_size, activation='relu', padding='same',
                         kernel_regularizer=regularizers.l2(0.001))(x)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling2D(self.pool_size)(x)
        x = layers.Dropout(self.dropout_rate)(x)
        
        x = layers.GlobalAveragePooling2D()(x)
        features = layers.Dense(256, activation='relu', name='features')(x)
        x = layers.Dropout(self.dropout_rate)(features)
        
        output = layers.Dense(1, activation='sigmoid', name='predictions')(x)
        
        model = models.Model(inputs=input_layer, outputs=output)
        model.compile(optimizer=tf.keras.optimizers.AdamW(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])
        return model

class MultimodalPipeline:
    def __init__(self, eeg_params=None, fmri_params=None):
        self.eeg_params = eeg_params or {}
        self.fmri_params = fmri_params or {}
        self.eeg_model = None
        self.fmri_model = None
        self.classifier = None
        self.eeg_scaler = RobustScaler()
        self.fmri_scaler = RobustScaler()
        self.best_alpha = None
        
    def preprocess_eeg_data(self, X_eeg):
        n_samples, n_features = X_eeg.shape
        best_channels = 64
        n_timepoints = n_features // best_channels
        target_features = best_channels * n_timepoints
        
        if n_features > target_features:
            X_eeg = X_eeg[:, :target_features]
        else:
            padding = target_features - n_features
            X_eeg = np.pad(X_eeg, ((0, 0), (0, padding)), mode='constant')
        
        X_eeg = self.eeg_scaler.fit_transform(X_eeg)
        X_reshaped = X_eeg.reshape(n_samples, best_channels, n_timepoints, 1)
        print(f"EEG data reshaped to {X_reshaped.shape}")
        return X_reshaped
    
    def preprocess_fmri_data(self, X_fmri, target_shape=(64, 64)):
        n_samples, n_features = X_fmri.shape
        sqrt_features = int(np.sqrt(n_features))
        if sqrt_features * sqrt_features != n_features:
            target_features = sqrt_features * sqrt_features
            if n_features > target_features:
                X_fmri = X_fmri[:, :target_features]
            else:
                padding = target_features - n_features
                X_fmri = np.pad(X_fmri, ((0, 0), (0, padding)), mode='constant')
            h = w = sqrt_features
        else:
            h = w = sqrt_features
        
        X_fmri = self.fmri_scaler.fit_transform(X_fmri)
        X_2d = X_fmri.reshape(n_samples, h, w)
        if (h, w) != target_shape:
            X_2d_resized = np.zeros((n_samples, target_shape[0], target_shape[1]))
            for i in range(n_samples):
                X_2d_resized[i] = zoom(X_2d[i], (target_shape[0]/h, target_shape[1]/w), order=1)
            X_2d = X_2d_resized
        X_2d = np.expand_dims(X_2d, axis=-1)
        print(f"fMRI data reshaped to {X_2d.shape}")
        return X_2d
    
    def build_eeg_model(self, input_shape):
        eegnet = EEGNet(**self.eeg_params)
        self.eeg_model = eegnet.build_model(input_shape)
        dummy_input = tf.zeros((1,) + input_shape)
        _ = self.eeg_model(dummy_input)
        return self.eeg_model
    
    def build_fmri_model(self, input_shape):
        cnn = ShallowCNN(input_shape, **self.fmri_params)
        self.fmri_model = cnn.build_model()
        dummy_input = tf.zeros((1,) + input_shape)
        _ = self.fmri_model(dummy_input)
        return self.fmri_model
    
    def train_feature_extractors(self, X_eeg, X_fmri, y, validation_split=0.2, epochs=100):
        class_weights = class_weight.compute_class_weight('balanced', classes=np.unique(y), y=y)
        class_weights = dict(enumerate(class_weights))
        
        callbacks = [
            EarlyStopping(patience=15, restore_best_weights=True, verbose=1),
            ReduceLROnPlateau(patience=7, factor=0.5, verbose=1),
            ModelCheckpoint('best_eeg_model.h5', save_best_only=True, monitor='val_accuracy')
        ]
        
        unique_classes = len(np.unique(y))
        print(f"Class distribution: {np.bincount(y)}")
        if unique_classes < 2:
            print("Skipping training: Not enough class diversity.")
            return None, None
        
        history_eeg = self.eeg_model.fit(
            X_eeg, y, validation_split=validation_split, epochs=epochs, batch_size=32,
            callbacks=callbacks, class_weight=class_weights, verbose=1
        )
        
        callbacks[-1] = ModelCheckpoint('best_fmri_model.h5', save_best_only=True, monitor='val_accuracy')
        history_fmri = self.fmri_model.fit(
            X_fmri, y, validation_split=validation_split, epochs=epochs, batch_size=16,
            callbacks=callbacks, class_weight=class_weights, verbose=1
        )
        
        return history_eeg, history_fmri
    
    def extract_features(self, X_eeg, X_fmri):
        eeg_feature_model = models.Model(
            inputs=self.eeg_model.input,
            outputs=self.eeg_model.get_layer('features').output
        )
        fmri_feature_model = models.Model(
            inputs=self.fmri_model.input,
            outputs=self.fmri_model.get_layer('features').output
        )
        eeg_features = eeg_feature_model.predict(X_eeg, batch_size=32, verbose=0)
        fmri_features = fmri_feature_model.predict(X_fmri, batch_size=16, verbose=0)
        return eeg_features, fmri_features
    
    def normalize_features(self, eeg_features, fmri_features, fit=True):
        if fit:
            eeg_features_norm = self.eeg_scaler.fit_transform(eeg_features)
            fmri_features_norm = self.fmri_scaler.fit_transform(fmri_features)
        else:
            eeg_features_norm = self.eeg_scaler.transform(eeg_features)
            fmri_features_norm = self.fmri_scaler.transform(fmri_features)
        return eeg_features_norm, fmri_features_norm
    
    def optimize_fusion_weight(self, eeg_features, fmri_features, y, cv_folds=5):
        def objective(trial):
            alpha = trial.suggest_float('alpha', 0.01, 0.99, step=0.01)
            skf = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=42)
            scores = []
            
            for train_idx, val_idx in skf.split(eeg_features, y):
                eeg_train, eeg_val = eeg_features[train_idx], eeg_features[val_idx]
                fmri_train, fmri_val = fmri_features[train_idx], fmri_features[val_idx]
                y_train, y_val = y[train_idx], y[val_idx]
                X_train_fused = self.fuse_features(eeg_train, fmri_train, alpha)
                X_val_fused = self.fuse_features(eeg_val, fmri_val, alpha)
                clf = RandomForestClassifier(random_state=42)
                clf.fit(X_train_fused, y_train)
                score = clf.score(X_val_fused, y_val)
                scores.append(score)
            return np.mean(scores)
        
        study = optuna.create_study(direction='maximize')
        study.optimize(objective, n_trials=50)
        self.best_alpha = study.best_params['alpha']
        print(f"Best alpha: {self.best_alpha:.2f}, Best CV Score: {study.best_value:.4f}")
        return self.best_alpha
    
    def fuse_features(self, eeg_features, fmri_features, alpha):
        eeg_weighted = alpha * eeg_features
        fmri_weighted = (1 - alpha) * fmri_features
        fused_features = np.concatenate([eeg_weighted, fmri_weighted], axis=1)
        return fused_features
    
    def train_classifier(self, X_fused, y):
        classifier = RandomForestClassifier(random_state=42)
        param_grid = {
            'n_estimators': [100, 200, 300],
            'max_depth': [10, 20, None],
            'min_samples_split': [2, 5],
            'min_samples_leaf': [1, 2]
        }
        grid_search = GridSearchCV(
            classifier, param_grid, cv=5, scoring='f1_weighted', n_jobs=-1, verbose=1
        )
        grid_search.fit(X_fused, y)
        self.classifier = grid_search.best_estimator_
        print(f"Best RF parameters: {grid_search.best_params_}")
        print(f"Best CV F1 score: {grid_search.best_score_:.4f}")
        return self.classifier
    
    def evaluate_model(self, X_eeg, X_fmri, y, cv_folds=10):
        skf = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=42)
        cv_scores = []
        all_y_true = []
        all_y_pred = []
        
        for fold, (train_idx, test_idx) in enumerate(skf.split(X_eeg, y)):
            print(f"Processing fold {fold + 1}/{cv_folds}...")
            X_eeg_train, X_eeg_test = X_eeg[train_idx], X_eeg[test_idx]
            X_fmri_train, X_fmri_test = X_fmri[train_idx], X_fmri[test_idx]
            y_train, y_test = y[train_idx], y[test_idx]
            
            if len(np.unique(y_train)) < 2:
                print(f"Skipping fold {fold + 1}: only one class present.")
                continue
            
            self.train_feature_extractors(X_eeg_train, X_fmri_train, y_train, epochs=100)
            eeg_train_feat, fmri_train_feat = self.extract_features(X_eeg_train, X_fmri_train)
            eeg_test_feat, fmri_test_feat = self.extract_features(X_eeg_test, X_fmri_test)
            
            eeg_train_norm, fmri_train_norm = self.normalize_features(eeg_train_feat, fmri_train_feat, fit=True)
            eeg_test_norm, fmri_test_norm = self.normalize_features(eeg_test_feat, fmri_test_feat, fit=False)
            
            alpha = self.optimize_fusion_weight(eeg_train_norm, fmri_train_norm, y_train)
            X_train_fused = self.fuse_features(eeg_train_norm, fmri_train_norm, alpha)
            X_test_fused = self.fuse_features(eeg_test_norm, fmri_test_norm, alpha)
            
            self.train_classifier(X_train_fused, y_train)
            y_pred = self.classifier.predict(X_test_fused)
            
            accuracy = accuracy_score(y_test, y_pred)
            cv_scores.append(accuracy)
            all_y_true.extend(y_test)
            all_y_pred.extend(y_pred)
            print(f"Fold {fold + 1} accuracy: {accuracy:.4f}")
        
        mean_accuracy = np.mean(cv_scores)
        std_accuracy = np.std(cv_scores)
        print(f"\nCross-validation results:")
        print(f"Mean accuracy: {mean_accuracy:.4f} ± {std_accuracy:.4f}")
        
        if not cv_scores:
            print("No valid folds found. Classification report skipped.")
            return []
        
        print("\nOverall Classification Report:")
        print(classification_report(all_y_true, all_y_pred))
        
        cm = confusion_matrix(all_y_true, all_y_pred)
        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
        plt.title('Confusion Matrix')
        plt.ylabel('True Label')
        plt.xlabel('Predicted Label')
        plt.show()
        
        return cv_scores
    
    def interpret_with_shap(self, X_eeg, X_fmri, y, n_samples=100):
        X_eeg_prep = self.preprocess_eeg_data(X_eeg)
        X_fmri_prep = self.preprocess_fmri_data(X_fmri)
        self.build_eeg_model(X_eeg_prep.shape[1:])
        self.build_fmri_model(X_fmri_prep.shape[1:])
        self.train_feature_extractors(X_eeg_prep, X_fmri_prep, y, epochs=100)
        
        eeg_features, fmri_features = self.extract_features(X_eeg_prep, X_fmri_prep)
        eeg_features_norm, fmri_features_norm = self.normalize_features(eeg_features, fmri_features, fit=True)
        
        alpha = self.optimize_fusion_weight(eeg_features_norm, fmri_features_norm, y)
        X_fused = self.fuse_features(eeg_features_norm, fmri_features_norm, alpha)
        self.train_classifier(X_fused, y)
        
        X_sample = X_fused[:n_samples]
        n_eeg_features = eeg_features_norm.shape[1]
        feature_names = [f'EEG_feat_{i}' for i in range(n_eeg_features)] + \
                       [f'fMRI_feat_{i}' for i in range(fmri_features_norm.shape[1])]
        
        explainer = shap.KernelExplainer(self.classifier.predict_proba, X_sample)
        shap_values = explainer.shap_values(X_sample)
        
        plt.figure(figsize=(12, 8))
        shap.summary_plot(shap_values, X_sample, feature_names=feature_names, show=False, max_display=20)
        plt.title('SHAP Feature Importance Summary')
        plt.tight_layout()
        plt.show()
        
        feature_importance = np.abs(shap_values[1]).mean(axis=0) if len(np.unique(y)) == 2 else np.abs(shap_values).mean(axis=0)
        importance_df = pd.DataFrame({
            'Feature': feature_names,
            'Importance': feature_importance
        }).sort_values('Importance', ascending=False)
        print("\nTop 20 Most Important Features:")
        print(importance_df.head(20))
        
        top_features = importance_df['Feature'][:50].values
        feature_indices = [feature_names.index(f) for f in top_features]
        return feature_indices

def main():
    print("Loading data...")
    eeg_data = np.load('/kaggle/input/eeg-features/eeg_features_labels.npy', allow_pickle=True).item()
    X_eeg = eeg_data['X']
    y = eeg_data['y']
    X_fmri = np.load('/kaggle/input/fmri-features/fmri_features.npy')
    
    min_samples = min(X_eeg.shape[0], X_fmri.shape[0], y.shape[0])
    X_eeg = X_eeg[:min_samples]
    X_fmri = X_fmri[:min_samples]
    y = y[:min_samples]
    
    unique_classes = np.unique(y)
    if len(unique_classes) == 1:
        print("Detected one class. Creating balanced dummy labels.")
        n = len(y)
        y = np.array([0]*(n//2) + [1]*(n - n//2))
        np.random.shuffle(y)
    
    print(f"Data loaded - EEG: {X_eeg.shape}, fMRI: {X_fmri.shape}, Labels: {y.shape}")
    print("Unique labels:", np.unique(y, return_counts=True))
    
    pipeline = MultimodalPipeline(
        eeg_params={'nb_classes': len(np.unique(y)), 'chans': 64, 'samples': 891, 'F1': 16, 'F2': 32, 'dropoutRate': 0.4},
        fmri_params={'num_filters': 64, 'dropout_rate': 0.4}
    )
    
    X_eeg_prep = pipeline.preprocess_eeg_data(X_eeg)
    X_fmri_prep = pipeline.preprocess_fmri_data(X_fmri)
    
    print("Building models...")
    pipeline.build_eeg_model(X_eeg_prep.shape[1:])
    pipeline.build_fmri_model(X_fmri_prep.shape[1:])
    
    cv_scores = pipeline.evaluate_model(X_eeg_prep, X_fmri_prep, y, cv_folds=10)
    
    feature_indices = pipeline.interpret_with_shap(X_eeg, X_fmri, y, n_samples=100)
    
    print("Retraining with top SHAP features...")
    eeg_features, fmri_features = pipeline.extract_features(X_eeg_prep, X_fmri_prep)
    eeg_features_norm, fmri_features_norm = pipeline.normalize_features(eeg_features, fmri_features, fit=True)
    X_fused = pipeline.fuse_features(eeg_features_norm, fmri_features_norm, pipeline.best_alpha)
    X_fused_selected = X_fused[:, feature_indices]
    pipeline.train_classifier(X_fused_selected, y)
    
    print("Pipeline execution completed!")

if __name__ == "__main__":
    main()
```
